# 04 — Baseline and 4-model comparison

70/15/15 stratified split → logistic regression baseline vs Random Forest, XGBoost, LightGBM. Scored by business-relevant metrics (F1, ROC-AUC, PR-AUC) — not raw accuracy.

```python
from src.models.pipeline import load_split_data
from src.models.model_zoo import build_zoo
from src.models.preprocessing import build_preprocessor
from src.evaluation.metrics import evaluate_model
import json

Xtr, Xva, Xte, ytr, yva, yte = load_split_data()
pre = build_preprocessor()
pre.fit(Xtr, ytr)

zoo = build_zoo()
results = {}
for name, model in zoo.items():
    model.fit(pre.transform(Xtr), ytr)
    results[name] = evaluate_model(model, pre, Xva, yva, name)[0]

for n, r in results.items():
    print(f"{n:30s} F1={r["f1"]:.4f} ROC={r["roc_auc"]:.4f} PR={r["pr_auc"]:.4f} Acc={r["accuracy"]:.4f}")
```

```python
json.dump(results, open("artifacts/metrics/model_comparison_metrics.json","w"), indent=1, default=str)
```
